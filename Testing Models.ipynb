# **Testing Model1**

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image
```

```python
IMG_SIZE = 224

# Fixed class meaning (VERY IMPORTANT)
CLASS_NAMES = {
    0: "Healthy (No Fall Armyworm)",
    1: "Diseased (Fall Armyworm)"
}
```

```python
model = load_model("model1_simple_cnn.h5")
print("Model loaded successfully")
```

```python
def predict_image(img_path):
    # Load & preprocess image
    img = image.load_img(img_path, target_size=(IMG_SIZE, IMG_SIZE))
    img_array = image.img_to_array(img) / 255.0
    img_array = np.expand_dims(img_array, axis=0)

    # Predict
    prob = model.predict(img_array)[0][0]
    label = 1 if prob >= 0.5 else 0
    confidence = prob if label == 1 else 1 - prob

    # Show image
    plt.imshow(img)
    plt.axis("off")
    plt.title(
        f"Prediction: {CLASS_NAMES[label]}\n"
        f"Confidence: {confidence:.2f}"
    )
    plt.show()

    return {
        "raw_probability": float(prob),
        "predicted_label": label,
        "confidence": float(confidence)
    }
```

```python
predict_image("Images/id_ybs49hkkd1snv.jpg")
predict_image("Images/id_ybzlndvtkabxy.jpg")
predict_image("Images/id_yexkuvbyapiua.jpg")
predict_image("Images/id_yfchdbopvmtev.jpg")
```

# **Testing Model4**

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image
from tensorflow.keras.applications.efficientnet import preprocess_input
```

```python
model4 = load_model("model4_efficientnet.h5")
```

```python
CLASS_NAMES = ["0", "1"]  # 0 = Healthy, 1 = Diseased
```

```python
IMG_SIZE = 224

def predict_image_model4(img_path):
    img = image.load_img(img_path, target_size=(IMG_SIZE, IMG_SIZE))
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)
    img_array = preprocess_input(img_array)

    pred = model4.predict(img_array)[0][0]

    label = CLASS_NAMES[1] if pred > 0.5 else CLASS_NAMES[0]
    confidence = pred if pred > 0.5 else 1 - pred

    plt.imshow(image.load_img(img_path))
    plt.axis("off")
    plt.title(f"Prediction: {label} | Confidence: {confidence:.2f}")
    plt.show()
```

```python
predict_image_model4("Images/id_xd5lnueuu2rpv.jpg")
predict_image_model4("Images/id_y7b2btlurju6c.jpg")
predict_image_model4("Images/id_y6xuj915mmy79.jpg")
predict_image_model4("Images/id_y5s59mdl0xlaj.jpg")
```
